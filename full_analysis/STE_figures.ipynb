```python
from lmfit import Parameters, Model
from numpy import exp


def logistic(x, mu, sigma, gamma, lambda_,):
    y = gamma + (1 - gamma - lambda_)*(1./(1+exp(-1*(sigma)*(x-mu))))
    return y

model = Model(logistic)


#model = StepModel(form='logistic')
params = Parameters()
params.add('mu', value = 50, min = .01, max = 99.99)
params.add('sigma', value = 1, min = 0, max = 1)
params.add('gamma', value = 0, min = 0, max = 0.5)
params.add('lambda_', value = 0, min = 0, max = 0.5);

x = np.arange(0,100,.1)
y = model.eval(params, x=x)

fig = plt.figure(figsize=[12.16/2.54, 10.32/2.54])
ax = fig.add_subplot()
ax.plot(x,y)
ax.set_ylim([0,1])
```
